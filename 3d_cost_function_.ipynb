```python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries ready!")
```

```python
# Creating range for w and b
w_range = np.linspace(-10, 10, 100)
b_range = np.linspace(-10, 10, 100)

# Creating a coordinate grid
W, B = np.meshgrid(w_range, b_range)

print("Grid generated!")
```

```python
# Soup bowl equation
Z = W**2 + B**2

print("Math calculated! Ready to plot.")
```

```python
# Setting up the 3D plotting environment
fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection='3d')

# Plotting the surface
surface = ax.plot_surface(W, B, Z, cmap='viridis', edgecolor='none')

# Adding labels
ax.set_xlabel('Weight (w)', labelpad=15)
ax.set_ylabel('Bias (b)', labelpad=15)
ax.set_zlabel('Cost J(w,b)', labelpad=15)
ax.set_title('Your Custom 3D Soup Bowl', pad=20)

# Adding a color bar side legend
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=5, pad=0.15)

plt.show()
```
